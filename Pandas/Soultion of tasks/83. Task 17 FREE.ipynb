###`Q-1:`
i. Create a DataFrame birds from the above dictionary data which has the index labels.

ii. Display basic information about the dataFrame.

iii. Show Alternate rows of the dataframe.

````python
import numpy as np
import pandas as pd
````

````python
# code here
data = {
    'birds': ['Cranes', 'Cranes', 'plovers', 'spoonbills', 'spoonbills', 'Cranes',
              'plovers', 'Cranes', 'spoonbills', 'spoonbills', 'Cranes'],
    'age': [3.5, 4, 1.5, np.nan, 6, 3, 5.5, np.nan, 8, 4, 3.5],
    'visits': [2, 4, 3, 4, 3, 4, 2, 2, 3, 2, 2],
    'priority': ['yes', 'yes', 'no', np.nan, 'no', 'no', 'no', 'yes', 'no', 'no', 'yes']
}

labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k']

# Create the DataFrame
df = pd.DataFrame(data, index=labels)

# Display the DataFrame
print(df)

# basic info about dataframe
print("\nBasic info about DataFrame\n")
df.info()

# Show Alternate rows of the dataframe
print("\nShow Alternate rows of the dataframe\n")
df[::2]
````

###`Q-3:`
i. Select all rows with nan values in age and visits column.

ii. Fill nan with respective series mode value.

````python
df
````

````python
# code here
print("i. Select all rows with nan values in age and visits column.\n\n",df[df[['age','visits']].isna().any(axis=1)])

# Fill NaN with mode value for each column
birds_filled = df.fillna(df.mode().iloc[0])
print("DataFrame after filling NaN with mode values:\n")
print(birds_filled)
````

###`Q-4`
i. Find the total number of visits of the bird Cranes

ii. Find the number of each type of birds in dataframe.

iii. Print no of duplicate rows

iv. Drop Duplicates rows and make this changes permanent. Show dataframe after changes.

````python
# code here
# 1. Find the total number of visits of the bird cranes
total_visits_cranes = df.loc[df['birds'] == 'Cranes', 'visits'].sum()
print("Total number of visits of Cranes:", total_visits_cranes)

# 2. Find the number of each type of birds in dataframe.
bird_counts = df['birds'].value_counts()
print("Number of each type of bird:\n", bird_counts)

# 3. Print number of duplicate rows
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

# 4. Drop duplicate rows permanently and show DataFrame
df.drop_duplicates(inplace=True)
print("\nDataFrame after dropping duplicate rows:")
print(df)

````
