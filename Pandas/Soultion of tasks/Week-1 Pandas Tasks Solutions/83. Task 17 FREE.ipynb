###`Q-6` Write a code which can display the bar chart of top 5 teams who have played maximum number of matches in the IPL.
>  Hint: Be careful the data is divided in 2 different cols(Team 1 and Team 2)


````python
# code here
import matplotlib.pyplot as plt
top_five_team1=ipl['Team1'].value_counts()
top_five_team2=ipl['Team2'].value_counts()
newTeam=top_five_team1+top_five_team2
x=newTeam.sort_values(ascending=False)
x=x.head(5)
plt.figure(figsize=(10,6))
plt.bar(x.index, x.values)
plt.title('Top 5 IPL Teams by Number of Matches Played (2008–2022)')
plt.xlabel('Teams')
plt.ylabel('Number of Matches')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
````

###`Q-8:` Team1 vs Team2 Dashbord
Create a function which will take two string(name of two teams) as input. Show win Loss record between them and player getting most player of the match award in matches between these two teams.
```
team1_vs_team2('Kolkata Knight Riders','Chennai Super Kings')
```

````python
# code here
def team1_vs_team2(team1, team2):
    # Filter matches where team1 and team2 played against each other
    matches = ipl[
        ((ipl['Team1'] == team1) & (ipl['Team2'] == team2)) |
        ((ipl['Team1'] == team2) & (ipl['Team2'] == team1))
    ]
    
    if matches.empty:
        print(f"No matches found between {team1} and {team2}")
        return
    
    # Count how many times each team won
    win_counts = matches['WinningTeam'].value_counts()
    
    # Get most frequent Player of the Match
    potm = matches['Player_of_Match'].value_counts().head(1)
    
    # Print win/loss summary
    print(f"Head-to-Head: {team1} vs {team2}")
    print(f"Total Matches Played: {len(matches)}")
    print("\nWin Summary:")
    for team, wins in win_counts.items():
        print(f"  {team}: {wins} wins")
    
    print("\n⭐ Most POTM in these matches:")
    print(f"  {potm.index[0]} - {potm.values[0]} awards")
    
    # Bar chart for visual clarity
    plt.figure(figsize=(6,4))
    plt.bar(win_counts.index, win_counts.values)
    plt.title(f'{team1} vs {team2} - Win Comparison')
    plt.xlabel('Teams')
    plt.ylabel('Number of Wins')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Example usage
team1_vs_team2('Kolkata Knight Riders', 'Chennai Super Kings')

````

### `Q-9:` Find out the top 7 cities where the matches of Kolkata Knight Riders are played frequently and plot the result as bar chart.

*`.plot(kind = "bar")` can help you to plot the bar chart. Also you can learn more about this method from [here](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.html)*



````python
# code here
matches_played_by_kkr=ipl[(ipl['Team1']=='Kolkata Knight Riders')|(ipl['Team2']=='Kolkata Knight Riders')]
plot=matches_played_by_kkr['City'].value_counts().head(7)

plt.figure(figsize=(10,8))
plt.bar(plot.index, plot.values)
plt.title('Cities where kkr played most')
plt.xlabel('Cities')
plt.ylabel('Number of matches')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
````
