# What is Generator
Python generators are simple way of creating iterators.

````python
# Simple example
# In generators there's no return statement but yield statement is there.
def gen_demo():
    yield "first statement"
    yield "second statement"
    yield "third statement"

gen=gen_demo() # returns generator
print(gen)
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
````

 **`yield` vs `return`**

---

### 🧠 1. Basic Idea

| Keyword      | What it does                                                                                       | Used in             |
| ------------ | -------------------------------------------------------------------------------------------------- | ------------------- |
| **`return`** | Ends the function and gives back **one final value**.                                              | Normal functions    |
| **`yield`**  | Pauses the function and gives back **a value temporarily**, but the function can **resume later**. | Generator functions |

---

### ⚙️ 2. Behavior Difference

#### 🟢 `return`

* When a function hits `return`, it **stops immediately**.
* You get **one** result.

Example:

```python
def get_numbers():
    return [1, 2, 3]

print(get_numbers())
```

**Output:**

```
[1, 2, 3]
```

➡️ All numbers are returned **at once**.

---

#### 🟣 `yield`

* When a function hits `yield`, it **pauses** and **remembers its state**.
* You can **resume** it later to get the next value.
* This creates a **generator** — an iterator that yields values one by one.

Example:

```python
def get_numbers():
    for i in range(1, 4):
        yield i

for num in get_numbers():
    print(num)
```

**Output:**

```
1
2
3
```

➡️ Numbers are produced **one at a time**, not all at once.

---

### ⚡ 3. Key Differences Summary

| Feature          | `return`                      | `yield`                               |
| ---------------- | ----------------------------- | ------------------------------------- |
| Stops function?  | Yes                           | Pauses (can resume)                   |
| Returns          | Single value                  | Generator (multiple values over time) |
| Memory use       | High (stores all data)        | Low (lazy evaluation)                 |
| Usage            | Regular function              | Generator function                    |
| Typical use case | Computations, data processing | Large data, streams, pipelines        |

---

### 💡 Example: Why `yield` is useful

Suppose you’re reading a huge file line by line — you don’t want to load the whole file in memory.

```python
def read_large_file(file_path):
    with open(file_path, 'r') as f:
        for line in f:
            yield line.strip()
```

This way, you process **one line at a time**, efficiently.

#### Range Function using Generator


````python
def my_range(start,end):
    for i in range(start,end):
        yield i

for i in my_range(100,151):
    print(i)
````

### Generator Expression(Similar to list comprehension)

````python
gen=(i**2 for i in range(1,51))

for i in gen:
    print(i)
````
