### What is an Iteration?
Iteration is a general term for taking each item of something , one after another. Any time you use a loop, explicit or implicit, to go over a group of items, that is iteration.


```python
# Example
num=[1,2,3]
for i in num:
    print(i)
```

### What is Iterator
An iterator is an object that allows the prorammer to reverse through a sequence of data without having to store the entire data in the memory.


```python
# Example
import sys
L=[x for x in range(1,10000)]
# for i in L:
#     print(i*2)

print(f"{sys.getsizeof(L)/64}MB Consumed")

# By this method the numbers are not stored in the memory. Working is same but memory doesn't occupied at all.
x=range(1,10000) # x is itertor here.
# for i in x:
#     print(x*2)
print(f"{sys.getsizeof(x)/64}MB Consumed")
```

### What is Iterable
- Iterable is an object, which one can iterate over.
- It generates an iterator when passed to iter() method.

```python
# Example
L=[1,2,3] # this L is an iterable
type(L) # gives list

iter(L) # returns list iterator
print(type(iter(L)))
```

### Understanding how for loop works

```python
num=[1,2,3]

for i in num:
    print(i)
```

```python
num=[1,2,3]
# step-1: fetch iterator using the iter()
iter_num=iter(num)
# Step-2: while using iter() calls next()
next(iter_num)# prints 1
next(iter_num)# prints 2
next(iter_num)# prints 3
next(iter_num)# raises an stopIteration Exception.
```

### Making our own for loop

```python
def custom_loop(iterable):
    iterator=iter(iterable)
    while(True):
        try:
            print(next(iterator))
        except StopIteration:
            break
```

```python
a=[1,2,3] # List
b=range(1,11) # range function
c=(1,2,3) # tuple
d={1,2,3} # set
e={0:1,1:2,2:3} # dictionary

custom_loop(a)
custom_loop(b)
custom_loop(c)
custom_loop(d)
custom_loop(e)
```

### Let's create our own range() function

```python
class my_range:
    def __init__(self,start,end):
        self.start=start
        self.end=end
    def __iter__(self):
        return my_range_iterator(self)
```

```python
class my_range_iterator:
    def __init__(self,iterableObj):
        self.iterable=iterableObj
    def __iter__(self):
        return self
    def __next__(self):
        if self.iterable.start>=self.iterable.end:
            raise StopIteration
        
        current=self.iterable.start
        self.iterable.start+=1
        return current        
```

```python
for i in my_range(1,11):
    print(i)
```
