###`Q-1:` Create a random 3x4 matrix with value between 0-100. And perform below tasks

    i. Sort this matrix. np.sort()
    ii. Sort this matrix based on values in 2nd column.
    iii. Sort this matrix based on max value in each row.
    iv. Sort based on elements value.


See examples:
```
arr =
    [[92 90 74]
    [ 6 63 93]
    [15 93 96]
    [70 60 48]]
```
```
i. np.sort
[[74 90 92]
 [ 6 63 93]
 [15 93 96]
 [48 60 70]]

ii. based on 2nd column
[[70 60 48]
 [ 6 63 93]
 [92 90 74]
 [15 93 96]]

iii. based on row max- ascending
[[15 93 96]
 [ 6 63 93]
 [92 90 74]
 [70 60 48]]

iv. based on elements value
[[ 6 15 48]
 [60 63 70]
 [74 90 92]
 [93 93 96]]

```

````python
# code here
import numpy as np

# Create random 4x3 matrix with values between 0–100
arr = np.random.randint(0, 101, (4, 3))
print("Original Array:\n", arr)

# i. Sort each row
sorted_arr = np.sort(arr, axis=1)
print("\n(i) Sorted each row:\n", sorted_arr)

# ii. Sort based on 2nd column (column index 1)
sorted_by_col2 = arr[arr[:, 1].argsort()]
print("\n(ii) Sorted by 2nd column:\n", sorted_by_col2)

# iii. Sort based on max value in each row
sorted_by_row_max = arr[np.max(arr, axis=1).argsort()]
print("\n(iii) Sorted by max value in each row:\n", sorted_by_row_max)

# iv. Sort based on all element values (flatten and reshape)
sorted_by_elements = np.sort(arr, axis=None).reshape(arr.shape)
print("\n(iv) Sorted by all elements:\n", sorted_by_elements)

````

###`Q-2:` There is an array of marks of 5 students in 4 subjects. Further you are asked to perform below task.
    i. Add marks every student of an extra subject in the same array.
    ii. Add two new students marks in respective 5 subjects.(one subject added in above task)
    iii. Add extra column with sum of all subjects(5-subjects) marks
    iv. Sort the array(non-ascending order) on total marks column--one added in above task. Show top 2 rows.

Note: Change dimension of arrays during concatenation or appending if required.

Given Array-
```
marks = [[13, 10,  9, 33],
       [63, 46, 90, 42],
       [39, 76, 13, 29],
       [82,  9, 29, 78],
       [67, 61, 59, 36]]

extra_subject = [41, 87, 72, 36, 92]
#Two extra students record-
rec1 = [77, 83, 98, 95, 89]
rec2 = [92, 71, 52, 61, 53]
```

````python
# code here
 

# Given data
marks = np.array([
    [13, 10,  9, 33],
    [63, 46, 90, 42],
    [39, 76, 13, 29],
    [82,  9, 29, 78],
    [67, 61, 59, 36]
])

extra_subject = np.array([41, 87, 72, 36, 92])
rec1 = np.array([77, 83, 98, 95, 89])
rec2 = np.array([92, 71, 52, 61, 53])

# i. Add marks of extra subject as a new column
marks = np.column_stack((marks, extra_subject))
print("After adding extra subject:\n", marks)

# ii. Add two new students (rows)
marks = np.vstack((marks, rec1, rec2))
print("\nAfter adding two new students:\n", marks)

# iii. Add extra column for total marks (sum across subjects)
total_marks = marks.sum(axis=1)
marks = np.column_stack((marks, total_marks))
print("\nAfter adding total marks column:\n", marks)

# iv. Sort in non-ascending order (by total marks column)
# (Total marks is now the last column, index = -1)
sorted_marks = marks[marks[:, -1].argsort()[::-1]]

print("\nSorted by total marks (descending):\n", sorted_marks)

# Show top 2 rows
print("\nTop 2 students:\n", sorted_marks[:2])
````

### `Q-5:` Get row numbers of NumPy array having element larger than X.
```
arr = [[1,2,3,4,5],
      [10,-3,30,4,5],
      [3,2,5,-4,5],
      [9,7,3,6,5]]

X = 6
```

````python
# code here
arr = np.array([
    [1, 2, 3, 4, 5],
    [10, -3, 30, 4, 5],
    [3, 2, 5, -4, 5],
    [9, 7, 3, 6, 5]
])

X = 6

# Find row indices where any element > X
rows = np.where(np.any(arr > X, axis=1))[0]

print("Row numbers having element larger than", X, ":\n", rows)
````

### `Q-6:` How to convert an array of arrays into a flat 1d array?


````python
# These arrays are given.
arr1 = np.arange(3)
arr2 = np.arange(3,7)
arr3 = np.arange(7,10)
````

````python
# code here
arr1 = np.arange(3)       # [0 1 2]
arr2 = np.arange(3, 7)    # [3 4 5 6]
arr3 = np.arange(7, 10)   # [7 8 9]

# Combine and flatten into one 1D array
flat_arr = np.concatenate((arr1, arr2, arr3))

print("Flattened 1D array:\n", flat_arr)
````

### `Q-7:` You are given a array. You have to find the minimum and maximum array element and remove that from the array.

```python
import numpy as np

np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))
```

````python
# code here
import numpy as np

# Given data
np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))

print("Original Array:\n", arr)

# Flatten to 1D for easy manipulation
arr_flat = arr.flatten()

# Find min and max values
min_val = np.min(arr_flat)
max_val = np.max(arr_flat)

print("\nMinimum value:", min_val)
print("Maximum value:", max_val)

# Remove min and max values
arr_removed = arr_flat[(arr_flat != min_val) & (arr_flat != max_val)]

print("\nArray after removing min and max elements:\n", arr_removed)
````

### `Q-8:` You are given an arrays. You have to limit this array's elements between 100 to 200. $arr ∈ [100, 700]$. So replace those values accordingly with the minimum and maximum value. Then sort the array and perform the cumulative sum of that array.

````python
# code here
import numpy as np

# Given array
np.random.seed(42)
arr = np.random.randint(100, 700, 20)
print("Original Array:\n", arr)

# 1️⃣ Limit array elements between 100 and 200
arr_clipped = np.clip(arr, 100, 200)
print("\nArray after limiting values between 100–200:\n", arr_clipped)

# 2️⃣ Sort the array
arr_sorted = np.sort(arr_clipped)
print("\nSorted Array:\n", arr_sorted)

# 3️⃣ Cumulative sum of sorted array
arr_cumsum = np.cumsum(arr_sorted)
print("\nCumulative Sum:\n", arr_cumsum)
````

### `Q-9:` You are given a array ($arr ∈ [0, 1]$). First you have round off the elements upto 3 decimal places and compare that
- 0th percentile == minimum value of the array
- 100th percentile == maximum value of the array
- also find the difference betwen 51th percenile and 50th percentile values

````python
# code here
import numpy as np

# Given array (values between 0 and 1)
np.random.seed(10)
arr = np.random.random(20)
print("Original Array:\n", arr)

# 1️⃣ Round elements to 3 decimal places
arr = np.round(arr, 3)
print("\nRounded Array:\n", arr)

# 2️⃣ Check percentile relationships
p0 = np.percentile(arr, 0)
p100 = np.percentile(arr, 100)

print("\n0th Percentile:", p0)
print("Minimum value :", np.min(arr))
print("100th Percentile:", p100)
print("Maximum value  :", np.max(arr))

print("\n0th Percentile == Min Value →", np.isclose(p0, np.min(arr)))
print("100th Percentile == Max Value →", np.isclose(p100, np.max(arr)))

# 3️⃣ Find difference between 51st and 50th percentiles
p51 = np.percentile(arr, 51)
p50 = np.percentile(arr, 50)
diff = p51 - p50

print("\n51st Percentile:", p51)
print("50th Percentile:", p50)
print("Difference (p51 - p50):", diff)
````
