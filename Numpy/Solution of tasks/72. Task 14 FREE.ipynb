###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
# code here
# Given array
arr = np.array([[1, 2, np.nan],
                [4, 2, 6],
                [np.nan, np.nan, 5]])

# Step 1: Remove all NaN values to find valid numbers
valid_values = arr[~np.isnan(arr)]

# Step 2: Find the most common number (mode)
# bincount works only with integers, so convert to int
counts = np.bincount(valid_values.astype(int))
most_common = np.argmax(counts)

# Step 3: Replace NaN values with the most common number
arr[np.isnan(arr)] = most_common

# Step 4: Print results
print("Most common element:", most_common)
print("Array after filling NaN values:\n", arr)
````

###`Q-6:` Given two arrays, X and Y, construct the Cauchy matrix C.
`Cij =1/(xi - yj)`

http://en.wikipedia.org/wiki/Cauchy_matrix
```
x = numpy.array([1,2,3,4]).reshape((-1, 1)
y = numpy.array([5,6,7])
```

````python
# code here
x = np.array([1, 2, 3, 4]).reshape(-1, 1)   # column vector
y = np.array([5, 6, 7])                     # row vector

# Formula: C[i, j] = 1 / (x[i] - y[j])
C = 1 / (x - y)

print("Cauchy Matrix:\n", C)
````

###`Q-7:` Plot this below equation.

$$\large y = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$

**Note: This equation is called tanh activation function. In deep learning, many times this function is used. If you find some difference between the sigmoid function and this tanh function, note that to your notebook.**

````python
# code here
import matplotlib.pyplot as plt
# Generate x values from -10 to 10
x = np.linspace(-10, 10, 200)

# Calculate y using the tanh formula
y = (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))

# Plot
plt.plot(x, y, label='tanh(x)', color='green')
plt.title('Tanh Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
````

### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
# code here
def check_broadcast(shape_a, shape_b):
    # Start from the last dimension and move backwards
    for a, b in zip(shape_a[::-1], shape_b[::-1]):
        if a != b and a != 1 and b != 1:
            return False
    return True

# Given shapes
shape_a = (3, 2, 2)
shape_b = (2, 2)

# Check if broadcastable
print(check_broadcast(shape_a, shape_b))
````
